```python
%%writefile lab2a.l
%option noyywrap

%{
#include <stdio.h>

int octalToDecimal(char *octal)
{
    int decimal = 0;
    int i;

    for (i = 0; octal[i] != '\0'; i++)
        decimal = decimal * 8 + (octal[i] - '0');

    return decimal;
}

int isValidOctal(char *number)
{
    int i;

    for (i = 0; number[i] != '\0'; i++) {
        if (number[i] < '0' || number[i] > '7')
            return 0;
    }

    return 1;
}
%}

%%
[0-9]+ {
          if (isValidOctal(yytext))
              printf("Octal %s = Decimal %d\n",
                     yytext, octalToDecimal(yytext));
          else
              printf("%s is not a valid octal number\n", yytext);
       }

[ \t\n]+ ;
.        ;
%%

int main(void)
{
    printf("Enter octal number(s):\n");
    yylex();
    return 0;
}
```

2(b)

```python
%%writefile lab2b.l
%option noyywrap

%{
#include <stdio.h>

int vowels = 0;
int consonants = 0;
int characters = 0;
int words = 0;
int lines = 0;

void countLetters(char *text)
{
    int i;

    for (i = 0; text[i] != '\0'; i++) {
        if (text[i] == 'a' || text[i] == 'e' ||
            text[i] == 'i' || text[i] == 'o' ||
            text[i] == 'u' || text[i] == 'A' ||
            text[i] == 'E' || text[i] == 'I' ||
            text[i] == 'O' || text[i] == 'U')
            vowels++;
        else
            consonants++;
    }
}
%}

%%
[a-zA-Z]+ {
             words++;
             characters += yyleng;
             countLetters(yytext);
          }

\n        {
             lines++;
             characters++;
          }

.         {
             characters++;
          }
%%

int main(int argc, char *argv[])
{
    if (argc != 2) {
        printf("Usage: ./lab2b filename.txt\n");
        return 1;
    }

    yyin = fopen(argv[1], "r");

    if (yyin == NULL) {
        printf("Cannot open file: %s\n", argv[1]);
        return 1;
    }

    yylex();
    fclose(yyin);

    printf("Number of vowels     : %d\n", vowels);
    printf("Number of consonants : %d\n", consonants);
    printf("Number of characters : %d\n", characters);
    printf("Number of words      : %d\n", words);
    printf("Number of lines      : %d\n", lines);

    return 0;
}
```
